# Data Manipulation with Pandas and NumPy 

Data manipulation is the process of changing, organizing, or transforming data to make it more useful, readable, and suitable for analysis. It involves taks like cleaning, filtering, sorting, grouping, or calculating new values from existing data. 

Tools like Pandas and NumPy, popular Python libraries, are often used to streamline these tasks. Pandas excels at handling structured data, like tables of crypto trades, for filtering or grouping, while NumPy supports fast numerical computations, such as calculating average prices or returns. 

For instance, a crypto investor might manipulate data by filtering trades to show only Bitcoin transactions over $5000, sorting them by timestamp to track price movements, and aggregating daily totals to assess trading volume. This helps uncover trends, optimize strategies, and ensures data accuracy, ultimately supporting better decisionsin the fast-paced crypto market.

### Pandas 

Pandas is a popular Python Library used for working with data. It helps you load, clean, analyze, and manipulate data easily. Think of it like an excel spreadsheet in Python! It's great for handling tables of data. 

Key Concepts: 

- Series: A single column of data (like a list of labels)

- Index: Labels for rows, helping you identify and access data

You can load data from files (CSV, Excel, e.t.c..), manipulate it, and save back. 

#### Installing/Importing pandas 

```python
import pandas as pd 
```

## Create a DataFrame 

A dataframe is like a table with rows and columns. Let's create one with sample cryptocurrency data(e.g. coin names, prices, and trade volumes)

```python
# Create a dictionary with crypto data 

data = {
    'Coin': ['Bitcoin', 'Ethereum', 'Ripple', 'Litecoin'],
    'Price': [120000, 3000, 0.85, 120],
    'Volume': [15000, 8000, 200000, 50000]
}

df = pd.DataFrame(data)

print(df)
```

#### Explore Data 

```python
df
```

```python
print(df.head(2))
```

```python
print(df.tail(2))
```

```python
print(df.info())
```

### Basic Data Manipulation 

```python
# Select a specific column 

print(df[['Coin', 'Price']])
```

```python
print(df['Coin'])
```

### Filtering Rows

```python
# Price greater than 

print(df[df['Price'] > 100])
```

### Sort Data 

Sort by a column like price in ascending or descending order 

```python
print(df.sort_values('Price'))  # Sort by price in ascending order 
```

```python
print(df.sort_values('Price', ascending=False))  # Sort in descending order 
```

### Add new column 

```python
# What will be our price after 2% change? 

df['Price_after_2Pct'] = df['Price'] * 1.02
print(df)
```

```python
df.describe()
```

### Save and Load data with Pandas 

```python
df.to_csv('crypto_data.csv', index=False)
```

## NumPy 

NumPy is a powerful python library for numerical computations, especially useful for working with arrays (list of numbers) and performing fast mathematical operations. It's a foundation for data manipulation, often used alongside Pandas, and is great for tasks like calculations, statistics, and handling crypto-related data. Below, i'll walk you through the basics step-by-step- installing NumPy, creating arrays, and performing common operations with simple code. 

#### Installing and Importing NumPy 

```python
import numpy as np
```

#### Create a NumPy Array

A NumPy array is like a list, but optimized for math and faster operations. Let's create arrays with sample crypto data (e.g., coin prices and trade volumes)

```python
prices = np.array([45000, 50000, 40000, 20000])
print(prices.shape)
print(prices)
```

#### Data Type

```python
print(prices.dtype)
```

### Basic Data Manipulation 

```python
prices = np.array([45000, 50000, 40000, 20000])

prices_plus_1000 = prices + 1000        #Add 1000 to all prices 

print(prices_plus_1000)
```

```python
# Compute for 2% increase in price 

price_s = np.array([46000, 51000, 41000, 21000])

price_increases = price_s * 1.02

print(price_increases)

```

```python
# Filter data 

higher_price = price_s[price_s > 2000]
print(higher_price)
```

### Basic Statistical Functions 

```python
volume = np.array([500000, 300000, 200000, 2500000])

print("\nStatistics for Volume")

print("Mean (Average): ", np.mean(volume))  # Average Volume 
print("Minimum: ", np.min(volume))  # Minimum Volume 
print("Maximum: ", np.max(volume))   # Maximum Volume 
print("Sum: ", np.sum(volume))   # Sum of Volume 
print("Standard Dev (Volatility): ", np.std(volume))   # Standard Dev (Volatility) of Volume 
```

#### Working with a 2D array 

NumPy can handle multi-dimensional arrays. Let's combine our data into a 2D array (like a Table)

```python
# Create 2D Array

crypto_data = np.array([
    [45000, 5000],
    [30000, 28000],
    [11998, 8943]
])

print(crypto_data)
```

```python
# Access a row 

crypto_data[2]
```

```python
# Access Column 

crypto_data[:,0]
```
