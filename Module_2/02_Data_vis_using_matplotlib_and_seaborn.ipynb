## Install/Import Matplotlib



```python
import matplotlib.pyplot as plt 
```

```python
student = ['daniel', 'alfred', 'amah', 'divine', 'ruth', 'george']
scores = [80, 90, 70, 60, 85, 74]
```

#### Line Plot 


```python
plt.plot(student, scores, color = 'red')

plt.title('Student Test Score')
plt.xlabel('Students')
plt.ylabel('Student Score')
plt.show()
```

##### Scatter Plot 

```python
plt.scatter(student, scores, color = 'green')

plt.title('Student Test Score')
plt.xlabel('Name')
plt.ylabel('Scores')
plt.show()
```

#### Bar Chart 

```python
plt.bar(student, scores, color = 'blue')

plt.title('Student Test Score')
plt.xlabel('Name')
plt.ylabel('Scores')
plt.show()
```

#### Histogram 

```python
plt.hist(scores, color = 'purple')

plt.title('Student Test Score')
#plt.xlabel('Name')
plt.xlabel('Scores')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(4,2))

plt.bar(student, scores, color = 'blue')

plt.title('Student Test Score')
plt.xlabel('Name')
plt.ylabel('Scores')
plt.show()
```

## Seaborn

Seaborn is a super easy tool in Python that helps you draw pretty pictures (called plots) of your data. it's like a magic crayon box built on top of Matplotlib, making graphs look nice and simple to create for 

### Basic Setup 

```python
import seaborn as sns
import pandas as pd
```

```python
student = ['daniel', 'alfred', 'amah', 'divine', 'ruth', 'george']
scores = [80, 90, 70, 60, 85, 74]

data = pd.DataFrame({'student' : student, 'scores' : scores})
print(data)
```

#### Line Plot 

```python
sns.lineplot(x='student', y='scores', data=data, color='blue')
plt.title('Student Test Scores')
plt.show()
```

#### Scatter Plot 

```python
sns.scatterplot(x='student', y='scores', data=data, color='red')

plt.title('Student Test Scores')
plt.show()
```

#### Bar Chart

```python
sns.barplot(x='student', y='scores', data=data, color='gray')

plt.title('Student Test Scores')
plt.show()
```

#### Histogram

```python
sns.histplot(x='scores', data=data, color='purple')

plt.title('How Scores Pile')
plt.show()
```
