```python
# ...existing code...

from selenium import webdriver
from selenium.webdriver.chrome.options import Options
import time
from bs4 import BeautifulSoup

# Set up headless Chrome (optional)
options = Options()
options.add_argument("--headless")

driver = webdriver.Chrome(options=options)
driver.get("https://app.artemisanalytics.com/chains?selectedChains=abstract%2Cacala%2Caioz%2Calgorand%2Captos%2Carbitrum")
time.sleep(7)  # Wait for JS to load content

soup = BeautifulSoup(driver.page_source, "html.parser")
print(soup.prettify()[:200])  # Print first 200 chars for inspection

driver.quit()
# ...existing code...
```

```python
# ...existing code...

# Example: Extract all tables
tables = soup.find_all("table")
for table in tables:
    print(table.prettify()[:300])  # Print first 300 chars of each table

# Example: Extract all divs with a specific class (replace 'your-class-name' with the actual class)
divs = soup.find_all("div", class_="your-class-name")
for div in divs:
    print(div.prettify()[:300])  # Print first 300 chars of each div

# Example: Extract all divs (not recommended if there are too many)
all_divs = soup.find_all("div")
print(f"Found {len(all_divs)} divs.") 

# ...existing code... 
```
