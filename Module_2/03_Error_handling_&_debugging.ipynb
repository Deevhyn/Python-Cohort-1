## Error Handling and Debugging 

### Error Handling 

A Python mechanism to manage and respond to errors (exceptions) that occur during program execution, preventing crashes. 

Purposes: Ensures programs handle unexpected issues (e.g., invalid input, file not found) gracefully, improving reliability and user experience. 

### Exception (Try/Except)

try: Run the code 

except: if it crashes, ignore error silently or do something else 

finally: to clean up no matter what happens 

```python
# Error Handling using Try/Except 

try:
    # Attempt to open a file that may not exist
    with open('non_existent_file.txt', 'r') as file:
        content = file.read()
except:
    pass

finally:
    # This block will always execute, regardless of whether an error occurred
    print("This will always run, even if an error occurrs.")
```

## Common Errors 

ValueError: When you give the right type of data but it doesn't make sense (like asking for a negative age)

TypeError: Mixing incompatible things (like adding text to a number)

FileNotFound: Trying to open a file that doesn't exist 

KeyError: Looking for a non-existent dictionary key 

```python
number = 'ten'

value = int(number)  # Attempt to convert a string to an integer
print(value)
```

```python
## ValueError Handling 

number = 'ten'

try: 
    value = int(number)  # Attempt to convert a string to an integer
except ValueError:
    print("caught a TypeError: cannot add string and integer")
```

```python
result = 5 + '5'  # Attempt to add an integer and a string
print(result)
```

```python
## TypeError Handling

try:
    result = 5 + '5'  # Attempt to add an integer and a string
except TypeError:
    print("caught a TypeError: cannot add string and integer")
```

```python
with open('another_non_existent_file.txt', 'r') as file:
    content = file.read()
print(content)
```

```python
## FileNotFound Error Handling

try:
    with open('another_non_existent_file.txt', 'r') as file:
        content = file.read()
except FileNotFoundError:
    print("File not found. Please check the file path.")
```

```python
my_dict = {'name': 'Alice', 'age': 30}
value = my_dict['address']
print(value)
```

```python
# KeyError Handling

my_dict = {'name': 'Alice', 'age': 30}
try:
    value = my_dict['address']  # Attempt to access a non-existent key
except KeyError:
    print("KeyError: The key 'address' does not exist in the dictionary.")
```

```python
my_list = [1, 2, 3]

value = my_list[5]
print(value)
```

```python
# IndexError Handling
my_list = [1, 2, 3]

try:
    value = my_list[5]  # Attempt to access an index that doesn't exist
except IndexError:
    print("IndexError: The index 5 is out of range for the list.")
```
